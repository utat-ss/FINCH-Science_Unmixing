```python
import pickle as pkl
import numpy as np
```

```python
cnn_dict = pkl.load(open('cnn_dict.pkl', 'rb'))
```

```python
for i in cnn_dict.keys():
    print(f"For {i}, Total R2: {cnn_dict[i]['metrics_total'][0][3]}, param#: {cnn_dict[i]['metrics_total'][0][0]}")

print("\n")

for i in cnn_dict.keys():
    print(f"For {i}, GV R2: {cnn_dict[i]['metrics_total'][0][4]}, param#: {cnn_dict[i]['metrics_total'][0][0]}")

print("\n")

for i in cnn_dict.keys():
    print(f"For {i}, NPV R2: {cnn_dict[i]['metrics_total'][0][5]}, param#: {cnn_dict[i]['metrics_total'][0][0]}")

print("\n")

for i in cnn_dict.keys():
    print(f"For {i}, Soil R2: {cnn_dict[i]['metrics_total'][0][6]}, param#: {cnn_dict[i]['metrics_total'][0][0]}")

print("\n")

for i in cnn_dict.keys():
    print(f"For {i}, RMSE: {np.sqrt(cnn_dict[i]['metrics_total'][0][2])}, param#: {cnn_dict[i]['metrics_total'][0][0]}")
```
